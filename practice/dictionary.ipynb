```python
#Q1. Nested Dictionary Update
#Given the following dictionary:
#students = {
#    &quot;S1&quot;: {&quot;Python&quot;: 70, &quot;SQL&quot;: 60, &quot;PowerBI&quot;: 40},
#    &quot;S2&quot;: {&quot;Python&quot;: 85, &quot;SQL&quot;: 70, &quot;PowerBI&quot;: 30},
#   &quot;S3&quot;: {&quot;Python&quot;: 60, &quot;SQL&quot;: 55, &quot;PowerBI&quot;: 50}
#}
#Increase the Python marks by 5 only for students whose total marks are less
#than 180.
#Modify the dictionary in-place.
    
d={
    "S1": {"Python": 70, "SQL": 60, "PowerBI": 40},
    "S2": {"Python": 85, "SQL": 70, "PowerBI": 30},
    "S3": {"Python": 60, "SQL": 55, "PowerBI": 50}
}
for i,j in d.items():
    t=sum(j.values())
    if t<180:
        j["Python"] += 5
print(d)
```

```python
#Q2. Sales Aggregation
#Given sales data:
#sales = [
#   {&quot;product&quot;: &quot;Laptop&quot;, &quot;amount&quot;: 50000},
#    {&quot;product&quot;: &quot;Mobile&quot;, &quot;amount&quot;: 20000},
#    {&quot;product&quot;: &quot;Laptop&quot;, &quot;amount&quot;: 55000},
#   {&quot;product&quot;: &quot;Tablet&quot;, &quot;amount&quot;: 15000},
#    {&quot;product&quot;: &quot;Mobile&quot;, &quot;amount&quot;: 25000}
#]
#Write a program to calculate total sales per product using a dictionary.

sales =[
    {"product": "Laptop", "amount": 50000},
    {"product": "Mobile", "amount": 20000},
    {"product": "Laptop", "amount": 55000},
    {"product": "Tablet", "amount": 15000},
    {"product": "Mobile", "amount": 25000}]
d={}
for i in sales:
    p=i["product"]
    a=i["amount"]
    if p not in d:
        d[p]=0
    d[p]+=a     
print(d) 
```

```python
#Q3. Given a list of numbers:
#nums = [10, 20, 30, 40, 50]
#Create a dictionary where:
#- Keys are indices
#- Values are the average of the last three elements (including current)
#- If fewer than three elements exist, take average of available elements

l=[10,20,30,40,50]
d={}
for key in range(len(l)):
    if key==0:
        avg_value=l[key]
    elif key==1:
        avg_value=(l[key]+l[key-1])/2
    else:
        avg_value=(l[key]+l[key-1]+l[key-2])/3
    d[key]=avg_value
print(d)
```
