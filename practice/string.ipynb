```python
#index() - return position of character.
#strobj.index(char/substring,[start],[stop])
#by defult start-0,stop-end of the string. (n-1)
s="welcome to bhopal"
print(s.index("o"))
print(s.index("o",5))
print(s.index("bhopal"))
print(s.index("o",5,9))
```

```python
#strobj.find(char/substring,[start],[stop])
s="welcome to bhopal"
print(s.find("o"))
print(s.find("o",5))
print(s.find("bhopal"))
print(s.find("o",5,9))
```
