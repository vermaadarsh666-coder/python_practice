```python
n=int(input("enter any number:"))
sum_even=0
sum_odd=0
i=1
while i<(2*n)+1:
    if i%2==0:
        sum_even=sum_even+i
    else:
        sum_odd=sum_odd+i
    i=i+1
print(sum_even)
print(sum_odd)
```

```python
#wap to print reverse of any given number.
n=int(input("enter any number:"))
rev=0
while n!=0:
    dig=n%10
    rev=(rev*10)+dig
    n=n//10
print(rev)
```

```python
#wap to check wheather a give number is armstrong or not .
n=int(input("enter any number:"))
a=len(str(n))
old=n
sm=0
while n!=0:
    dig=n%10
    sm=sm+(dig**a)
    n=n//10
if sm==old:
    print(old,"armstrong")
else:
    print(old,"armstrong")
```
