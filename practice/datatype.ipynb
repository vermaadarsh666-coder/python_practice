```python
#list-collection of elements. 
#it is denoted by[] with csv. 
#ordered,intexed,mutable and heterogenous in 
#nature. 
l=[] #empty list
print(l)
print(type(l))
```

```python
l=[10,20,30,"hello",10.5]
print(l)
print(type(l))
print(l[0])  #indexed 
l[3]="decode" # mutable
print(l)
```

```python
#collection of elements
#denoted by () with csv.
#orderd,indexed,immutable and heterogenous in nature.
t=()  #empty tuple
print(t)
print(type(t))
```

```python
#string valued tuple
t=(10,)
print(t)
print(type(t))
```

```python
t=(10,20,30,"hello",10.5)
print(t)
print(type(t))
print(t[0])  #indexed
t[0]=1000    #immutable
print(t)
```

```python
#dictionary
#collection of element
#denoted by {k:v} and data stored in key :value pair
#keys must be unique and value can be duplicate
#ordered,unindexed and mutable. 
d={}  #empty dictionary
print(d)
print(type(d))
```

```python
d={"name":"abc","subject":"da","fees":30000}
print(d)
print(d["subject"])  #unindexed
d["Name"]="abc"    #mutable and values can be duplicate
d["name"]="decode"  #keys must be unique
print(d)
```

```python
#set  - collection  of unique element.
#denoted by {} with csv
#unordered,unindexed,mutable and heterogenous
s=set()    #empty set and set constructore
print(s)
print(type(s))
```

```python
s={10,20,30,"hello",10.5,10,20,10,20,10}
print(s)
print(type(s))
s.add(1000)
print(s)
```

```python
#frozenset - collection of unique element 
#denoted by constructor i.e frozenset()
#unordered,indexed,immutable and heterogenous
f=frozenset()
print(f)
print(type(f))
```

```python
f=frozenset({10,20,30,"hello",10.5,10,20,10,10})
print(f)
print(type(f))
```
