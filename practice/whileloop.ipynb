```python
#wap to print sum of even number and odd number.
n=int(input("enter any number :"))
sum_even=0
sum_odd=0
for i in range (1,(2*n)+1):
    if i%2==0:
        sum_even=sum_even+i
    else:
        sum_odd=sum_odd+i
        print("sum of first",n,"even_number:",sum_even)
        print("sum of first",n,"odd_number:",sum_odd)
```
