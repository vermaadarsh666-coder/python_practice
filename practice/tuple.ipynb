```python
# function 
# len(tupleobject)
t=(10,20,30,40,"hello",10.5)
print(len(t))
count=0
for i in t:
    count=count+1
print(count)
```

```python
# max()
# min()
t=(10,20,30,40,10.5)
print(max(t))
print(min(t))
```

```python
t=(10,20,30,40,10.5)
max_value=t[0]
for i in t:
    if i>max_value:
        max_value=i
print(max_value)
```

```python
t=(10,20,30,40,10.5)
min_value=t[0]
for i in t:
    if i<min_value:
        min_value=i
print(min_value)
```

```python
#sorted
# sorted(seqtobject,[reverse=false]) asc.
t=(11,10,1,2,16,9)
print(sorted(t))
```

```python
t = (10, 1, 19, 2)
l = list(t)
for i in range(len(l)):
    for j in range(i + 1, len(l)):
        if l[i] < l[j]:
            l[i], l[j] = l[j], l[i]   # swap
print(tuple(l))
```

```python
# count()
# tupleobj.count(element)
t=(10,20,30,10,40,20,10,10.10,20.0,20,30)
print(t.count(100))
print(t.count(10))
```

```python
# index()
# tupleobj.index(element,[start],[stop])
t=(10,20,30,10,40,"hello","h",100,200,10)
print(t.index(10))
print(t.index(10,1))
print(t.index(10,4))
print(t.index(300))
```
