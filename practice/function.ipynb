```python
def leap_year(year):
    if year%4==0:
        if year%100==0:
            if year%400==0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False
```

```python
x=int(input("enter any year:"))
if leap_year(x):
    print(x,"leap year")
else:
    print(x,"not a leap year")
```

```python
# wap function that return weather a give number is prime or not.
def prime_number(n):
    if n==0 or n==1:
        return False
    elif n>1:
        for i in range(2,n):
            if n%i==0:
                return False
                break
            else:
                return True
    else:
        return False
```

```python
x=int(input("enter any number:"))
if prime_number(x):
    print(x,"prime number")
else:
    print(x,"not a prime")
```

```python
#wap function that return weather a give number armstrong or not.
def arm_strong(n):
    a=len(str(n))
    old=n
    sm=0
    while n!=0:
        dig=n%10
        sm=sm+(dig**a)
        n=n//10
    return sm==old
```

```python
x=int(input("enter any number: "))
if arm_strong(x):
    print(x,"armstrong")
else:
    print(x,"not an armstrong")
```

```python
#wap function that return mean,median,mode a given list.
def mean_median_mode(l):
    sum,count=0,0
    for i in l:
        sum=sum+i
        count=count+1
    mean=sum/count
    sorted_list=sorted(l)
    n=len(l)
    if n%2==0:
        median=(sorted_list[n//2-1]+sorted_list[n//2])/2
    else:
        median=sorted_list[n//2]
    d={}
    for i in l:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    max_value=max(d.values())
    for i,j in d.items():
        if j==max_value:
            mode=i
    return mean,median,mode
```

```python
#wap function that return list asc order or desc depand on the input parameter take 2 parameter
# 1) list obj
# 2) optical parameter reverse by default reverse=false.
def sort_list(l1,reverse=False):
    l=list(l1)
    if reverse==False:
        for i in range(len(l)):
            for j in range(i+1,len(l)):
                if l[i]>l[j]:
                    l[i],l[j]=l[j],l[i]
        return l
    else:
        for i in range(len(l)):
            for j in range(i+1,len(l)):
                if l[i]<l[j]:
                    l[i],l[j]=l[j],l[i]
        return l
```

```python
#2)#wap function that return weather a give number armstrong or not.(range function).
def armstrong_range(start, end):
    for n in range(start, end + 1):
        if n > 0:
            old = n
            a = len(str(n))
            sum = 0

            while old > 0:
                digit = old % 10
                sum += digit ** a
                old = old // 10

            if sum == n:
                print(n, end=" ")
```

```python
start = int(input("enter start: "))
end = int(input("enter end: "))

armstrong_range(start, end)
```

```python
#wap function that return largest among 3 number.
def max_of_two(x,y):
    if x>y:
        return x
    return y
```

```python
def max_of_three(a,b,c):
    return max_of_two(a,max_of_two(b,c))
```

```python
max_of_three(10,20,30)
```
