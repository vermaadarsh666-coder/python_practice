```python
#syntax:-
# lambda  agrument : expression 
```

```python
#ex-
a=lambda x:x**2
print(a(5))
```

```python
a=lambda x,y:x+y
print(a(10,20))
```

```python
a=lambda a,b,c:a if (a>b and a>c) else b if b>c else c
print(a(10,8,12))
```

```python
l=[1,2,3,5]
#[1,8,27,64,125]
l1=[]
for i in l:
    l1.append(i**3)
print(l)
print([i**3 for i in l])
print(list(map(lambda i:i**3,l)))

```

```python
l1=[5,1,6,2,3,4]
l2=[3,6,1,5,4,2]
# (125,1,6,32,81,16)
l3=[]
for i,j in zip(l1,l2):
    l3.append(i**j)
print(tuple(l3))
print(tuple([i**j for i,j in zip(l1,l2)]))
print(tuple(map(lambda i,j:i**j,l1,l2)))
```

```python
l=[[100,200,50,45],[1,7,5,7],[10,30,40]]
#[100,7,30]
l1=[]
for i in l:
    for j in range(len(i)):
        for k in range(j+1,len(i)):
            if i[j]>i[k]:
                i[j],i[k]=i[k],i[j]
    l1.append(i[-2])
print(l1)
print([sorted(i)[-2] for i in  l])
print(list(map(lambda i:sorted(i)[-2],l)))
```
