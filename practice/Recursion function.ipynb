```python
# write a recursive function sum of first n natural number.
def sum (n):
    if n==1:
        return n 
    else:
        return n+sum(n-1)
```

```python
sum(10)
```

```python
#write a recursive function factoral of give number 
def fact(n):
    if n==0 or n==1:
        return 1 
    else:
        return n*fact(n-1)
```

```python
fact(1)
```

```python
def fibo(n):
    if n==1:
        return 0
    elif n==2:
         return 1
    else:
         return fibo(n-1)+fibo(n-2)
```

```python
fibo(2)
```

```python
n=int(input("enter any number:"))
fact=1
for i in range (1,n+1):
    fact=fact*i
print(fact)
```
